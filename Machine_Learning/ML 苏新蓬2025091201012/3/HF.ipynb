```python
from transformers import pipeline

# 加载情感分析管道（自动使用镜像下载模型）
classifier = pipeline("sentiment-analysis", model="hfl/chinese-roberta-wwm-ext")

# 要分析的中文句子
texts = [
    "这家餐厅的味道太棒了，下次还来",
    "今天天气很糟糕，心情也不好"
]

# 进行情感分析并打印结果
for text in texts:
    result = classifier(text)[0]
    print(f"文本：{text}")
    print(f"情感：{result['label']}，置信度：{result['score']:.2f}\n")
```

```python
# 要分析的中文句子
texts = [
    "这家餐厅的味道太棒了，下次还来",
    "今天天气很糟糕，心情也不好",
    "damn,too bad!"
]

# 进行情感分析并打印结果
for text in texts:
    result = classifier(text)[0]
    print(f"文本：{text}")
    print(f"情感：{result['label']}，置信度：{result['score']:.2f}\n")
```

一开始很奇怪这里的置信度怎么才0.5出头。（后来又运行了一遍又来到0.8）。豆包说，因为这个模型只是认识中文字符，还没有情感分析的功能。他的判断依据是这里的情感都是label-1，而不是消极积极，所以没有训练好
